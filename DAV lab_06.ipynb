```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio

pio.templates.default="plotly_white"
print("Plotly version:",__import__("plotly").__version__)
```

```python
df = pd.read_csv("esports_player_performance_tournament_analytics.csv")
print("Shape:", df.shape)
df.head()
```

```python
fig1 = px.histogram(
    df,
    x="performance_score",
    nbins=20,
    title="Distribution of player performance scores",
    labels={"performance_score": "Performance score"},
)

fig1.update_layout(
    yaxis_title="Number of players",
    xaxis_title="Performance score",
    bargap=0.05,
    template="plotly_white",
)

from IPython.display import HTML, display
display(HTML(fig1.to_html()))
```

```python
MVP_ORDER = ["No", "Yes"]
MVP_COLORS = {"No": "#7f8c8d", "Yes": "#2ecc71"}

plot_df = df.copy()
plot_df = plot_df.dropna(subset=["accuracy_percent", "reaction_time_ms", "mvp_award"]).copy()

fig2 = px.scatter(
    plot_df,
    x="accuracy_percent",
    y="reaction_time_ms",
    color="mvp_award",
    category_orders={"mvp_award": MVP_ORDER},
    color_discrete_map=MVP_COLORS,
    hover_name="team_name",
    hover_data={
        "player_role": True,
        "map_played": True,
        "match_type": True,
        "performance_score": True,
        "mvp_award": False,
    },
    opacity=0.7,
    labels={
        "accuracy_percent": "Accuracy (%)",
        "reaction_time_ms": "Reaction time (ms)",
        "mvp_award": "MVP award",
    },
    title="Accuracy vs. reaction time, coloured by MVP award",
)

fig2.update_layout(template="plotly_white")
from IPython.display import HTML, display
display(HTML(fig2.to_html()))
```

```python
role_summary = (
    df.groupby("player_role", as_index=False)["performance_score"].mean()
    .sort_values("performance_score", ascending=False)
)

fig3 = px.bar(
    role_summary,
    x="player_role",
    y="performance_score",
    color="player_role",
    title="Average performance score by player role",
    labels={
        "player_role": "Player role",
        "performance_score": "Average performance score",
    },
    text_auto=".2f",
)

fig3.update_layout(template="plotly_white", xaxis_title="Player role", yaxis_title="Average performance score")
from IPython.display import HTML, display
display(HTML(fig3.to_html()))
```
